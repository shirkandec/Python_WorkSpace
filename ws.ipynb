```python
# Takes a string as input
# Counts how many vowels (a, e, i, o, u) are present
# Prints the total number of vowels


a=input("Enter the string ")
count=0
for char in a:
    if char in"aeiou":
        count+=1

print("this is count of vowels in given string:",count)
```

```python
# Count consonants
# Input a string and count the number of consonants.

a=input("Enter ths string")
count=0
for char in a:
    if char not in "aeiou":
        count+=1
print("this is count of consonants in given string:",count)
```

```python
# Count uppercase and lowercase letters
# Input a string and display the number of uppercase and lowercase characters.

a=input("Enter the string")
upper=0
lower=0
for char in a:
    if char.isupper():
        upper+=1
    elif char.islower:
        lower+=1

print ("count of uppercase",upper)
print("count of lowercase",lower)        
```

```python
# Check whether a string is a palindrome
# Same idea as your program, but make it work without using [::-1].
a=input("Enter the string")
reverse=""
for char in range(len(a)-1,-1,-1):
    reverse = reverse + a[char]
if a.casefold()==reverse.casefold():
    print("paliendrome")
else:
    print("not paliendrome")

    
    
```
