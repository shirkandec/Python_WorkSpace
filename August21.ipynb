```python
l1 = [1, 2, 3, 4]
l2 = [7, 8, 9, 0]
```

```python
print("l1 + l2 =", l1 + l2)
```

```python

print("l1 - l2 ", [a - b for a, b in zip(l1, l2)])
```

```python
print("l1 * l2 ", [a * b for a, b in zip(l1, l2)])
```

```python
print("l1 - 5 =", [a - 5 for a in l1])
```

```python
print("l1 + 5 =", [a + 5 for a in l1])
```

```python
print( l1 * 5)
```

```python
print("l1 > l2 =", l1 > l2)
```
