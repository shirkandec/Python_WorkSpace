```python
list1=[[10,20,30,40],[50,60],[70,80,90]]
```

```python
print(list1)
```

```python
list2=[]
for i in list1:
    for values in i :
        list2.append(values)

print(list2)
```

```python
l2=[10,[20,30,40],77,[50,60,70,80,90],[100,110]]
print (l2)
```

```python
l3=[]
for i in  l2:
    if type(i)==int:
        l3.append(i)
        continue
     
    for val in i:
        l3.append(val)


print(l3)
        
           
```

```python
l4=[]
for i in l2:
    if isinstance(i,list):
        l4.extend(i)
    elif isinstance(i,int):
        l4.append(i)

print(l4)
```
