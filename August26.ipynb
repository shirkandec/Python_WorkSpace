```python
numbers = [12, 45, 7, 23, 45, 89, 12, 34, 5, 67]
```

```python
# 6. Print all even numbers.
for i in numbers:
    if i%2==0:
        print(i)
```

```python
# 7. Print all odd numbers.
for i in numbers:
    if i%2==1:
        print(i)
```

```python
# 8. Count how many even and odd numbers are present.
count1=0
count2=0
for i in numbers:
    if i%2==0:
        count1+=1
    if i%2==1:
        count2+=1

print(count1)
print(count2)
```
