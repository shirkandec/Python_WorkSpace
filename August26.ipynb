```python
students = ["Rahul", "Priya", "Amit", "Sneha", "Raj", "Anjali"]
```

```python
9# 26. Print all student names.
print(students)
```

```python
# 27. Print names that start with "A".
for name in students:
    if name.startswith("A"):
        print (name)
```

```python
# 28. Print names containing more than five characters.
for name in students:
    if len(name)>5:
        print(name)
```

```python
# 29. Convert every name to uppercase.
for name in students:
    uppercase = name.upper()
    print(uppercase)
```

```python
# 30. Sort the names alphabetically.
for i in range(len(students)):


    for j in range(i+1, len(students)):

        
        if students[i] > students[j]:
            students[i], students[j] = students[j], students[i]
print(students)
```

```python
# 31. Find the longest student name.
longrst_name = ""
for name in students:
    if len(name)>len(longest_name):
        longest_name=name
print(longest_name)
```

```python
# 32. Check whether "Sneha" exists in the list
for name in students:
    if name == "Sneha":
        print("Sneha exists in the list")
```

```python
# 33. Create a new list containing the length of every name.
new_list=[]
for name in students:
    new_list.append(len(name))
print(new_list)
```

```python
# 34. Remove "Raj" from the list.
for name in students:
    if name=="raj":
        students.remove(name)
print(students)
```

```python
# 35. Print all names in reverse alphabetical order.
for i in range(len(students)):


    for j in range(i+1, len(students)):

        
        if students[i] > students[j]:
            students[i], students[j] = students[j], students[i]

students.reverse()
print(students)
```

```python
# 37. Find common elements between two lists.
l1 = [1, 2, 3, 4, 5]
l2 = [4, 5, 6, 7, 8]
common_elements = []
for i in l1:
    if i in l2:
        common_elements.append(i)
print(common_elements)
```

```python
# 39. Rotate a list one position to the right.
for i in range(len(l1)-1, 0, -1):
    l1[i], l1[i-1] = l1[i-1], l1[i]
print(l1)
```
