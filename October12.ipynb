```python
db = {1: [78, 97, 86], 2: [97, 94, 92], 3: [78, 90, 98]}

topper_roll = list(db.keys())[0]
topper_marks = sum(db[topper_roll]) / len(db[topper_roll])

for k, v in db.items():
    avg_marks = sum(v) / len(v)

    if avg_marks > topper_marks:
        topper_roll = k
        topper_marks = avg_marks



print(f"Roll number {topper_roll} is topper with average marks {topper_marks}. Congratulations!!!!!")
```

```python
db = {
    1: [78, 97, 86],
    2: [97, 94, 92],
    3: [78, 90, 98]
}

subjects = ["Physics", "Chemistry", "Maths"]

for i in range(len(subjects)):
    topper_roll = list(db.keys())[0]
    topper_marks = db[topper_roll][i]

    for k, v in db.items():
        if v[i] > topper_marks:
            topper_roll = k
            topper_marks = v[i]

    print(f"{subjects[i]} topper is roll number {topper_roll} with marks {topper_marks}")
```
