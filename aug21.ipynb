# Q. How to access data from tuple ?
## -> Indexing & Slicing 

```python
category = ("Electronics" , "Furniture" , "Electronics" , "Clothing")
print(category)
# Typecasting of data 
# Convert tuple into list -> do changes --> List -> Tuple 
cat_list = list(category)
print(cat_list)
cat_list.append("Stationary") # type: ignore
print(cat_list)
category = tuple(cat_list)
print(category)
print(id(category))
print(id(category[0]))
print(id(category[1]))
print(id(category[2]))

```

## Packing & Unpacking of elements / data 

```python
bag1 = ("Clothes" , "Charger", "Cash" , "whatever")  # tuple 
bag2 = ["Clothes" , "Charger", "Cash" , "whatever"]  # List

print(bag1)
v1,*v2 = bag1   # Unpacking 
print(v1 , v2)

b1,b2,b3,b4 = bag2
print(b1,b2,b3,b4)
```

```python
t1 = "A"
t2 = 20
t3 = "B"
t4 = "Z"

bag1 = t1,t2,t3,t4   # Packing of elements
print(bag1)
print(type(bag1))
```
