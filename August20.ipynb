```python
product_mrp=[10000,3000,5000,15000,18000,12000,3000,9000,20000]
selling_p=[]
for i in product_mrp:
    if i>=15000:
        a=i*(15/100)
        c=i-a
        selling_p.append(c)
        

    elif i>=10000:
        d=i*(10/100)
        e=i-d
        selling_p.append(e)
    elif i>=5000:
        f=i*(5/100)
        g=i-f
        selling_p.append(g)
    elif i<5000:
        selling_p.append(i)
    

print(selling_p)
```
