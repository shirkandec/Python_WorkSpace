```python

customer_name = input("Enter your name: ")
number_of_product = int(input("Enter the number of product purchased: "))
total_bill = 0

for i in range(number_of_product):
    product_name = input("Enter the product name: ")
    price = eval(input("Enter the price: "))
    quantity = int(input("Enter the Quantity: "))

    total_bill += price * quantity

print("Customer Name:", customer_name)
print("Total Bill:", total_bill)

if total_bill > 5000:
    dis_amount = total_bill * 10 / 100
    total_bill = total_bill - dis_amount
elif total_bill > 3000:
    dis_amount = total_bill * 5 / 100
    total_bill = total_bill - dis_amount
else:
    dis_amount = 0

if total_bill > 5000:
    extra_dis = total_bill - 200
    print("Final Bill:", extra_dis)
else:
    print("Final Bill:", total_bill)
```
