```python
movies = {
    'Avarapan 2': ["Emran Hasmi", "Disha Patani", "Ritesh Deshmukh"],
    'Dhurandhar 2': ["Ranbir Singh", "Akshay Khanna", "Sanjay Datta"],
    'Border 2': ["Sunny Deol", "Varun Dhawan", "Dilip Joshi"],
    'Ramayana': ["Ranbir Kapoor", "Sai Pallavi", "Yash"],
    'Spider-Man': ["Tom Holland", "Zendaya", "Jacob Batalon"]
}
```

```python
print(movies)
```

```python
for i in movies:
    print(i)
    
```

```python
for name in movies:
    if name == "Dhurandhar 2":
        count = 0

        for actor in movies[name]:
            for ch in actor.lower():
                if ch == 'a':
                    count += 1

print("Count of a in Dhurandhar 2 is", count)
```

```python
print(movies.keys())
nameip=input("Enter movie name: ")
ch_count=input("Enter character to count: ")
for name in movies:
    if name == nameip:
        count = 0

        for actor in movies[name]:
            for ch in actor.lower():
                if ch == ch_count.lower():
                    count += 1

print("Count of", ch_count, "in", nameip, "is", count)
```
