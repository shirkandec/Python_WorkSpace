# Conditional Statements

## if :->

## if - else :->
### ---> We use if - else to check one complete condition
### sty:->      if condition : 
###                     if block code
###             else :
###                     else block code
### when if condition does not satisfy then else block will be definately exicute

```python
rohan_marks = 75
sahil_marks = 82

if rohan_marks > sahil_marks:
    print("Rohan scored more")

elif rohan_marks == sahil_marks:
    print("Both scored the same marks")

else:
    print("Sahil scored more")

print("Final Result")
```

```python
day = "Sunday"
day2="saturday"
day3="Monday"

if day == "Saturday" and day == "Sunday" and day2 == "Saturday" and day2 == "Sunday" and day3 == "Saturday" and day3 == "Sunday":
    print("It's a weekend")
else:
    print("It's a weekday")
print("End of program")
```

## or :->
    If any operand is True then op is True.
    If all the operands are False then and then only op is false.

```python
attendence= "online" 
attendence= "offline"
if (attendence == "online" or attendence == "offline"):
    print("You are present")
else:
    print("You are absent")
```
