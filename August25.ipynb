```python
# print minimum & Second min 
l1=[58,67,66,56,45,43]
l2=[]

minimum=l1[0]
for i in l1:
    if i<minimum:
        minimum=i
for i in l1:
    if i != minimum:
        l2.append(i)

second_minimum = l2[0]

for i in l2:
    if i < second_minimum:
        second_minimum = i


print(second_minimum)
print(minimum)
```
