```python
employees = {
101: {"name": "Amit", "department": "IT", "salary": 45000},
102: {"name": "Priya", "department": "HR", "salary": 38000},
103: {"name": "Sneha", "department": "IT", "salary": 52000},
104: {"name": "Raj", "department": "Sales", "salary": 41000},
105: {"name": "Anjali", "department": "HR", "salary": 47000}
}
```

```python
# 6. Print employee ID with employee name.
for emp_id, val in employees.items():
    print(emp_id, val["name"])
```

```python
# 7. Print employees who belong to the IT department.
for emp_id, val in employees.items():
    if val["department"] == "IT":
        print(emp_id, val["name"])
```

```python


# 8. Print employees whose salary is greater than 40,000.
for emp_id,val in employees.items():
    if val['salary']>40000:
        print(val["name"], val["salary"])
```

```python
# 9. Find the employee with the highest salary.
highest_sal = 0
emp_name = ""

for emp_id, val in employees.items():
    if val["salary"] > highest_sal :
        highest_sal  = val["salary"]
        emp_name = val["name"]

print("Employee:", emp_name)
print("Highest salary:", highest_sal )
```

```python
# 10. Calculate the average salary of all employee
total=0
count=0
for val in employees.values():
    total+= val["salary"]
    count+=1

avg=total/count
print(F"This is the average salary of the employee {avg}")
```

```python
products = {
"P101": {"name": "Laptop", "price": 55000, "stock": 8},
"P102": {"name": "Mouse", "price": 700, "stock": 25},
"P103": {"name": "Keyboard", "price": 1200, "stock": 15},
"P104": {"name": "Monitor", "price": 14000, "stock": 5},
"P105": {"name": "Headphones", "price": 2500, "stock": 0}
}
```

```python
# 11. Print the name and price of every product.
for k, v in products.items():
    print(v["name"], v["price"])
```

```python
# 12. Print products having stock less than 10.
for k, v in products.items():
    if v["stock"]<10:
        print(f"{v}")
    
```

```python
# 13. Print products that are out of stock.
for k, v in products.items():
    if v["stock"]<1:
        print(f"{v}")
```

```python
# 14. Calculate the total value of each product using price x stock.
for k, v in products.items():
    a=v["price"]*v["stock"]
    print(a)
```

```python
# 15. Apply a 5% discount to all products costing more than 10,000.
for k, v in products.items():
    if v["price"]>10000:
        v["price"]=v["price"]-v["price"]*5/100

for k, v in products.items():
    print(k, v["name"], v["price"])
```
