# Assignment operator.
### --> Assign right hand side value to the left hand side value with the help of assignment operator.

# Composite Assignment operators-->
### += , -= , *= , /= , //= , %= , **=

```python
c = 1       
d = c+5           
print(d)            
```

```python
a = 10           
a = a + 5
print(a)
```

```python
x = 10
x += 5
print(x)

y = 25
y -= 5
print(y)

z = 35
z *= 3
print(z)

p = 25
p /= 3
print(p)

q = 18
q //= 5
print(q)

r = 40
r %= 3
print(r)

s = 5
s **= 3
print(s)
```

### 2 Binary
### 3 Octal
### 4 Hexadecimal

## Decimal
### 0 to 9
### base 10 system


```python
a = 8
print(a)
print(type(a))
```

## Binary
### 0 to 1
### base 2 system


```python
num1 = 0b1010
num2 = 0b1010

print(num1)
print(num2)
print(type(num1))
```

```python
var = 0o1010
print(var)
```

# Bitwise operators
### & , | , ^ , ! , >> , << , >>>

```python


num1 = 12      # 1100
num2 = 10      # 1010

print(num1 & num2)   # 1000 = 8
```

```python


value1 = 9     # 1001
value2 = 6     # 0110

print(value1 | value2)   # 1111 = 15
```

```python


x = 14         # 1110
y = 9          # 1001

print(x ^ y)   # 0111 = 7
```

```python


number = 20    # 10100
shift = 2

print(number >> shift)   # 20 // 4 = 5
```

```python


first = 6      # 0110
left = 2

print(first << left)   # 6 × 4 = 24
```

```python


decimal_num = 11

print(format(decimal_num, '08b'))   # 00001011
```
